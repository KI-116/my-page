## 一、语法

### 1. tensor创建

tensor: 是PyTorch中最基本的数据结构，类似于numpy中的ndarray。tensor可以在GPU上进行计算。通常为数值型数据，支持多维数组。

```python
import torch

x = torch.tensor([1,2,3])   #创建[1,2,3]的一维tensor
x = torch.randn(2,3)        #创建2行3列的随机tensor
x = torch.zeros(2,3)        #创建2行3列的全零tensor
x = torch.ones(2,3)         #创建2行3列的全一tensor

torch.arange(10)            #创建一个从0到9的一维tensor
torch.arange(0, 10, 2)      #创建一个从0到9的步长为2的一维tensor

torch.eye(4)                #创建一个4行4列的单位矩阵tensor
```

### 2. shape操作

1. reshape: 改变tensor的形状，返回一个新的tensor，原tensor不变。

```python
x = torch.randn(2,3,4)  # 创建一个2行3列4通道的随机tensor

x.view(6,4)             # tensor是连续的
x.reshape(6,4)          # 如果无法返回 view，则自动创建新的连续张量并复制数据。
```

2. transpose

```python
x.transpose(1,2)    #转置tensor的维度，(1,2)表示将第1维和第2维交换位置
```

```python
Q = Q.view(B,S,H,d_k).transpose(1,2)

# (B,S,H,d_k)
# ->
# (B,H,S,d_k)
# B: batch size
# S: sequence length
# H: number of heads
# B保持不变是因为(1,2)表示将第1维和第2维交换位置，B是第0维，所以不变
```

3. contiguous: 返回一个连续的tensor，如果原tensor是连续的，则返回原tensor，否则返回一个新的连续tensor。

```python
x = x.transpose(1,2)

x = x.contiguous().view(...) #transpose 后内存不连续。
```

4. unsqueeze 
在指定位置插入一个维度为1的维度，返回一个新的tensor，原tensor不变。

```python
x = torch.randn(10)
print(x.unsqueeze(0).shape)  # (1,10)
```

```python
x = torch.randn(10)

x.unsqueeze(0)
# (1,10)

```

```python
x.unsqueeze(1)
# (10,1)
```

5. squeeze

```python
#删除长度为1的维度
x.squeeze()  
```

6. cat

arg1: tensor列表，要求所有tensor的shape除了指定的维度外，其余维度必须相同。
arg2: 指定的维度。

```python
K = torch.randn(2,3,4)
V = torch.randn(2,3,4)
K_prev = torch.randn(2,3,4)
V_prev = torch.randn(2,3,4)

K = torch.cat([K_prev, K], dim=2)
V = torch.cat([V_prev, V], dim=2)
print(K.shape)  # (2,3,8)
print(V.shape)  # (2,3,8)
```

## 数学计算API

1. softmax

```python
import torch.nn.functional as F
x = torch.tensor([[2.0, 1.0, 0.1], [1.0, 2.0, 3.0]])
F.softmax(x, dim=-1)    # sim = -1: 指定在最后一个维度上进行softmax计算
```

2. sigmoid,silu,etc.

```python
torch.sigmoid(x) #sigmoid函数
```

```python
x * torch.sigmoid(x) # silu函数
```

```python
F.gelu(x)
```

```python
F.logsigmoid(x) #DPO loss中使用的logsigmoid函数
```

```python
# mean
x.mean(dim=-1)
```
